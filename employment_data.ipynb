```python
#top5 local authorities by employment
# Step 1: Filter for 'Local Authority' and 'Region' since nation and aggregation are totals
filtered_df = df[df["Authority_Type"].isin(["Local Authority", "Region"])]

# Step 2: Group by Local_Authority_County and sum Total_Employment
grouped = filtered_df.groupby("Local_Authority_County")["Total_Employment"].sum()

# Step 3: Get top 5 by employment
top_5_la_region = grouped.nlargest(5)
print(top_5_la_region)
```

```python
#year by year employment growth
df_sorted = df.sort_values(["Local_Authority_County", "Year"])
df_sorted["prev_year_employment"] = df_sorted.groupby("Local_Authority_County")["Total_Employment"].shift(1)
df_sorted["yoy_growth_percent"] = round(
    (df_sorted["Total_Employment"] - df_sorted["prev_year_employment"]) * 100 /
    df_sorted["prev_year_employment"].replace(0, pd.NA), 2
)
```
